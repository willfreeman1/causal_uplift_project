# 03 — Honest grading (built before any CATE model)

A plan is just one offer (or none) per customer. We grade it on the **holdout**, where assignment is still random, by keeping the people who by luck got the offer the plan would have chosen and reweighting them.

This is **not** adding up the model's own uplift predictions. That circular grade comes later as a cautionary comparison, never as a headline.

```python
from causal_uplift.data import load_and_split
from causal_uplift.economics import ARM_MENS, ARM_WOMENS
from causal_uplift.grading import ate_minus_cost, run_sanity_checks

_, grade = load_and_split()
print(f"grade set: {len(grade):,} customers")
print(f"always men's expected (ATE − $0.50): {ate_minus_cost(grade, ARM_MENS):.4f}")
print(f"always women's expected (ATE − $0.20): {ate_minus_cost(grade, ARM_WOMENS):.4f}")
```

```python
table = run_sanity_checks(grade)
table
```

## What must be true

- **Treat nobody** scores exactly 0 extra profit.
- **Random** is about 0, inside its confidence range.
- **Always Men's / Women's** match that arm's measured spend effect minus its send cost.

If those hold, the evaluation backbone is sound. Models can come next.
