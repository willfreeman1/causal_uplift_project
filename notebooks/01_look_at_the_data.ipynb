# 01 — Look at the data before modeling

Hillstrom MineThatData email experiment: ~64,000 customers randomly assigned to Men's email, Women's email, or no email.

Assignment mix, covariate balance, and average effects before any model.

```python
from causal_uplift.data import load_and_split, load_hillstrom
from causal_uplift.eda import run_eda
```

## Assignment mix and covariate balance

If randomization worked, each arm is about one-third of the list and customer characteristics line up across arms.

```python
df = load_hillstrom()
build, grade = load_and_split()
print(f"full={len(df):,}  build={len(build):,}  grade={len(grade):,}")

results = run_eda(df)
results["arm_mix"]
```

```python
results["feature_balance"]
```

## Average effects vs no email

These are simple differences in means. They will look small. They hide who each email helps.

```python
results["average_effects"]
```

## The response pattern that matters

Men's email works across the board. Women's email only competes for customers who bought women's items last year. There is no slice here that reacts badly to contact — sleeping dogs are demonstrated later, on the made-up dataset.

```python
pattern = results["visit_response_pattern"]
pattern[pattern["group"].isin(["womens", "mens", "newbie", "recent"])].sort_values(["group", "value", "arm"])
```

## What we found

- The three arms really are balanced (each ~33.3%; standardized mean differences vs control are tiny).
- Men's email raised visits by about **7.7 percentage points**, purchases by about **0.68 points**, and spend by about **$0.77** per customer.
- Women's email: about **4.5 points** on visits, **0.31 points** on purchases, **$0.42** on spend. Real, but weaker.
- For last-year women's-item buyers, the two emails are close on visits. For everyone else, Women's email barely moves visits and Men's still does.
- No group we checked is hurt by contact.

That is why the budgeted plan will roughly be: cheap Women's email to women's-item buyers, expensive Men's email to other strong responders, skip people whose response is near zero.
